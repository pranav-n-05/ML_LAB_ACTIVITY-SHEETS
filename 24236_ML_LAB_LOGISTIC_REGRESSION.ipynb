# 6.6 Logistic Regression – Exercises

## Exercise 1: GRE, GPA and institution prestige vs. admission
Dataset: https://stats.idre.ucla.edu/stat/stata/dae/binary.dta

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn import metrics
import seaborn as sn, matplotlib.pyplot as plt

data = pd.read_stata('https://stats.idre.ucla.edu/stat/stata/dae/binary.dta')
print(data.shape)
data.head()
```

```python
print(data.isnull().sum())
x = pd.get_dummies(data[['gre', 'gpa', 'rank']], columns=['rank'], drop_first=True)  # rank (prestige 1-4) is categorical
y = data['admit']
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=5)

model = LogisticRegression(solver='lbfgs', max_iter=1000)
model.fit(x_train, y_train)
y_pred = model.predict(x_test)

print('Confusion Matrix :\n', metrics.confusion_matrix(y_test, y_pred))
print('Accuracy Score :', metrics.accuracy_score(y_test, y_pred))
print(metrics.classification_report(y_test, y_pred))
print(dict(zip(x.columns, model.coef_[0])))
```

```python
sn.heatmap(pd.crosstab(y_test, y_pred, rownames=['Actual'], colnames=['Predicted']), annot=True, fmt='d')
plt.show()
```

## Exercise 2: Is a person susceptible to a disease?
Dataset: DATAtab *Medical example (logistic regression)* – https://datatab.net/statistics-calculator/regression/logistic-regression-calculator?example=Medical_example_logistic_regression

Open the link, download the example data as CSV and upload it to Colab as `/content/medical_example.csv`.

```python
import pandas as pd
data = pd.read_csv('/content/medical_example.csv')
print(data.shape)
data.head()
```

```python
print(data.isnull().sum())
data = data.dropna()
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn import metrics
import seaborn as sn, matplotlib.pyplot as plt

# Assumption: last column is the target, all other columns are features (categoricals are one-hot encoded)
target = data.columns[-1]
x = pd.get_dummies(data.drop(columns=[target]), drop_first=True)
y = data[target]
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=5)
model = LogisticRegression(max_iter=5000)
model.fit(x_train, y_train)
y_pred = model.predict(x_test)
print('Accuracy:', metrics.accuracy_score(y_test, y_pred))
print(metrics.classification_report(y_test, y_pred))
sn.heatmap(pd.crosstab(y_test, y_pred, rownames=['Actual'], colnames=['Predicted']), annot=True, fmt='d')
plt.show()
```

## Exercise 3: Which product will a customer buy?
Dataset: DATAtab *logistic regression example* – https://datatab.net/statistics-calculator/regression/logistic-regression-calculator?example=logistic_regression

Download as CSV, upload to Colab as `/content/product_example.csv`. If the target has >2 products, `LogisticRegression` automatically fits a multinomial model.

```python
import pandas as pd
data = pd.read_csv('/content/product_example.csv')
print(data.shape)
data.head()
```

```python
print(data.isnull().sum())
data = data.dropna()
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn import metrics
import seaborn as sn, matplotlib.pyplot as plt

# Assumption: last column is the target, all other columns are features (categoricals are one-hot encoded)
target = data.columns[-1]
x = pd.get_dummies(data.drop(columns=[target]), drop_first=True)
y = data[target]
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=5)
model = LogisticRegression(max_iter=5000)
model.fit(x_train, y_train)
y_pred = model.predict(x_test)
print('Accuracy:', metrics.accuracy_score(y_test, y_pred))
print(metrics.classification_report(y_test, y_pred))
sn.heatmap(pd.crosstab(y_test, y_pred, rownames=['Actual'], colnames=['Predicted']), annot=True, fmt='d')
plt.show()
```
