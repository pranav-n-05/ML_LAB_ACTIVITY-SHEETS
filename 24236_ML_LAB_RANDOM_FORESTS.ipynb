## Exercise 1: Fruit basket – majority voting
Tree-1 → Apple (Class-A), Tree-2 → Apple (Class-A), …, Tree-n → Banana (Class-B). The forest outputs the class with the most votes.

```python
from collections import Counter

votes = ['Apple', 'Apple', 'Banana']      # Tree-1, Tree-2, ..., Tree-n
counts = Counter(votes)
print(counts)
print('Final class (majority vote):', counts.most_common(1)[0][0])
```

```python
# Same idea with a real random forest on a small fruit basket
import numpy as np, pandas as pd
from sklearn.ensemble import RandomForestClassifier

rng = np.random.default_rng(0)
apples  = pd.DataFrame({'weight': rng.normal(150, 15, 60), 'length': rng.normal(7, .5, 60),  'fruit': 'Apple'})
bananas = pd.DataFrame({'weight': rng.normal(120, 15, 60), 'length': rng.normal(18, 1.5, 60), 'fruit': 'Banana'})
basket = pd.concat([apples, bananas])
rf = RandomForestClassifier(n_estimators=25, random_state=0).fit(basket[['weight', 'length']], basket['fruit'])

sample = pd.DataFrame([[148, 7.2]], columns=['weight', 'length'])
tree_votes = [t.predict(sample.values)[0] for t in rf.estimators_]
tree_votes = [rf.classes_[int(v)] for v in tree_votes]
print(Counter(tree_votes))
print('Forest prediction:', rf.predict(sample)[0])
```

**Answer:** Two of the three shown trees vote Apple (Class-A) and one votes Banana (Class-B), so majority voting gives **Apple** – the fruit likely to be taken most often.
