## Exercise 1: Handwritten digit recognition with kNN
Dataset: GitHub `pbharrin/machinelearninginaction` (Ch02/digits – 32x32 text images).

```python
!git clone -q https://github.com/pbharrin/machinelearninginaction.git
!ls machinelearninginaction/Ch02/digits
```

```python
import os, numpy as np
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score, confusion_matrix

def img2vector(path):
    return np.array([int(c) for line in open(path).read().split() for c in line]).reshape(1024)

def load(folder):
    files = sorted(f for f in os.listdir(folder) if f.endswith('.txt'))
    X = np.array([img2vector(os.path.join(folder, f)) for f in files])
    y = np.array([int(f.split('_')[0]) for f in files])
    return X, y

base = 'machinelearninginaction/Ch02/digits/'
X_train, y_train = load(base + 'trainingDigits')
X_test,  y_test  = load(base + 'testDigits')
print(X_train.shape, X_test.shape)
```

```python
knn = KNeighborsClassifier(n_neighbors=3).fit(X_train, y_train)
pred = knn.predict(X_test)
print('Accuracy:', accuracy_score(y_test, pred))
print(confusion_matrix(y_test, pred))
```

```python
import matplotlib.pyplot as plt
i = 0
plt.imshow(X_test[i].reshape(32, 32), cmap='gray_r')
plt.title(f'Actual {y_test[i]}, predicted {pred[i]}'); plt.show()
```

## Exercise 2: Euclidean distance between a new entry and existing values
The PDF does not print the new entry; the standard example **Brightness = 20, Saturation = 35** is used (change `new` to test others).

```python
import numpy as np, pandas as pd

df = pd.DataFrame({'Brightness': [40, 50, 60, 10, 70, 60, 25],
                   'Saturation': [20, 50, 90, 25, 70, 10, 80],
                   'Class': ['Red', 'Blue', 'Blue', 'Red', 'Blue', 'Red', 'Blue']})
new = (20, 35)
df['Distance'] = np.sqrt((df.Brightness - new[0])**2 + (df.Saturation - new[1])**2)
df = df.sort_values('Distance').reset_index(drop=True)
df
```

```python
k = 5
print(df.head(k))
print('Predicted class for', new, 'with k =', k, ':', df.head(k).Class.value_counts().idxmax())
```
